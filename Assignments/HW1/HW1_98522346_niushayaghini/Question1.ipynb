The KDE plot is a technique that allows for non-parametric estimation of the probability density function of a random variable. Its primary use is to visually represent the distribution of a dataset, as well as to evaluate the assumptions made by parametric models and to uncover patterns and trends in the data.


You can plot a column of the dataframe using the following command in seaborn:
```
sns.kdeplot(df['column1'])
plt.show()
```
or mutiple columns using:
```
sns.kdeplot(df[['column1', 'column2', 'column3']])
```

A histogram is a statistical graph that represents the distribution of a continuous dataset through plotted bars, each representing a particular category or class interval. The bar height reflects the frequency or count of data points within each group.

1. Dream Weight VS Actual Weight

Do you see any limits?

````python
# PLotting with sns / bin = 20

# Plot histogram for Dream Weight
sns.histplot(df['Dream Weight'], bins=20, color='blue', kde=True, label='Dream Weight')

# Plot histogram for Actual Weight
sns.histplot(df['Actual Weight'], bins=20, color='red', kde=True, label='Actual Weight')

# Configure and display the plot
plt.legend()
plt.xlabel("Weight")
plt.ylabel("Frequency")
plt.title("Comparison of Dream Weight vs Actual Weight")
plt.show()
````

````python
# PLotting with sns / bin = 10

# Plot histogram for Dream Weight
sns.histplot(df['Dream Weight'], bins=10, color='blue', kde=True, label='Dream Weight')

# Plot histogram for Actual Weight
sns.histplot(df['Actual Weight'], bins=10, color='red', kde=True, label='Actual Weight')

# Configure and display the plot
plt.legend()
plt.xlabel("Weight")
plt.ylabel("Frequency")
plt.title("Comparison of Dream Weight vs Actual Weight")
plt.show()
````

````python
# Plotting using matplotlib

# Plot histogram for Dream Weight
plt.hist(df['Dream Weight'], bins=20, color='blue', edgecolor='black', alpha=0.5, label='Dream Weight')

# Plot histogram for Actual Weight
plt.hist(df['Actual Weight'], bins=20, color='red', edgecolor='black', alpha=0.5, label='Actual Weight')

# Configure and display the plot
plt.legend()
plt.xlabel("Weight")
plt.ylabel("Frequency")
plt.title("Comparison of Dream Weight vs Actual Weight")
plt.show()
````

There is a method for dataframes in pandas called .corr
<br/>
This method calculates the correlation matrix for all columns.
<br/>
search about different values for 'method' parameter in df.corr and use them.

````python
# Pearson correlation between two specific columns
pearson_corr = df['Dream Weight'].corr(df['Actual Weight'], method='pearson')
print("Pearson Correlation between Dream Weight and Actual Weight:", pearson_corr)
````

````python
# Pearson correlation: measures the linear relationship between two continuous variables
pearson_corr = df.corr(method='pearson')

# saving in a csv file
pearson_corr.to_csv('Q1_pearson_correlation_matrix.csv', index=True)

print("Pearson Correlation Matrix:\n", pearson_corr)
````

````python
# Kendall correlation: measures the ordinal association between two variables
kendall_corr = df.corr(method='kendall')

# saving in a csv file
kendall_corr.to_csv('Q1_kendall_correlation_matrix.csv', index=True)

print("\nKendall Correlation Matrix:\n", kendall_corr)
````

````python
# Spearman correlation: assesses the monotonic relationship between two variables
spearman_corr = df.corr(method='spearman')

# saving in a csv file
spearman_corr.to_csv('Q1_spearman_correlation_matrix.csv', index=True)

print("\nSpearman Correlation Matrix:\n", spearman_corr)
````
