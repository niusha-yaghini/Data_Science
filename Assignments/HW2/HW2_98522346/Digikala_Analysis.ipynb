# Part13

Write a code that downloads the images in the 'data_product_images_main_url' column automatically.
<br/>
Run the code for 10 random rows and then plot the images using matplotlib.


```python
# column information for data_product_images_main_url

DPIMU_unique = df['data_product_images_main_url'].unique()

print(f"Get the count of unique values in a column count: {DPIMU_unique.size}\n")

print("the value count:", df['data_product_images_main_url'].value_counts().head())

print("\namount of null values: ", null_count.data_product_images_main_url)
```

```python
# Select 10 non-null random rows from the original DataFrame
random_image_rows = df['data_product_images_main_url'].dropna().sample(n=min(10, len(df))).reset_index(drop=True)
random_image_rows
```

```python
# Extract valid URLs from the selected rows
valid_urls = []
for url_list in random_image_rows:
    try:
        # Convert the string to a Python list and extract the first URL
        urls = ast.literal_eval(url_list)
        if isinstance(urls, list) and len(urls) > 0 and urls[0].startswith("https://"):
            valid_urls.append(urls[0])  # Append the first valid URL
    except (ValueError, SyntaxError):
        continue  # Skip invalid rows
    
valid_urls
```

```python
# Download and display the images
fig, axes = plt.subplots(1, len(valid_urls), figsize=(15, 5))
for i, url in enumerate(valid_urls):
    try:
        # Download the image
        response = requests.get(url)
        img = plt.imread(BytesIO(response.content), format='jpeg')
        
        # Display the image
        axes[i].imshow(img)
        axes[i].axis('off')  # Remove axis ticks
        axes[i].set_title(f'Image {i+1}')
    except Exception as e:
        print(f"Error downloading image {i+1}: {e}")

plt.tight_layout()
plt.show()
```

# Part15

Make another column based on the column 'data_product_rating_rate,' that has 10 values, 1 to 5.
<br/>
Each rate value must be categorized into one of these 5 categories based on the following rules.
<br/>
<br/>
0<rate<20 ==> A
<br/>
20<rate<40 ==> B
<br/>
40<rate<60 ==> C
<br/>
60<rate<80 ==> D
<br/>
80<rate<100 ==> E


```python
# column information

DPRR = df['data_product_rating_rate']
DPRR_unique = DPRR.unique()
print(f"DPRR_unique: {DPRR_unique} \n")

DPRR_unique_count = DPRR_unique.size
print(f"Get the count of unique values in a column count: {DPRR_unique_count}\n")

# the third one is null
# print("the value count:", df['data_product_rating_rate'].value_counts())

print("\namount of null values: ", null_count.data_product_rating_rate)
```

```python
DPRR_categorised = df[['data_product_id', 'data_product_rating_rate']].copy()

DPRR_categorised['data_product_rating_rate'] = DPRR_categorised['data_product_rating_rate'].fillna(0)

# Define the bins and corresponding labels
bins = [0, 20, 40, 60, 80, 100]  # Define the ranges
labels = ['A', 'B', 'C', 'D', 'E']  # Define the category labels

# Categorize the data
DPRR_categorised['rating_category'] = pd.cut(DPRR_categorised['data_product_rating_rate'], bins=bins, labels=labels, right=False)

# Fill null categories (if necessary) with a custom label (e.g., '0')
DPRR_categorised['rating_category'] = DPRR_categorised['rating_category'].cat.add_categories('0').fillna('0')

# Keep only the required columns
DPRR_categories = DPRR_categorised[['data_product_id', 'rating_category']]

print(type(DPRR_categories))
DPRR_categories.head()
```
